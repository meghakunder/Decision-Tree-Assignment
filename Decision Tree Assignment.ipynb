## Decision Tree

**Q1. What is a Decision Tree? Explain its structure (Root, Nodes, Leaves) with a real-life example.**

A Decision Tree is a supervised machine learning algorithm used for **classification** and **regression**. It makes decisions by asking a series of questions and splitting the data into smaller groups until it reaches a final prediction.

**Structure of a Decision Tree**

**1. Root Node**

The Root Node is the first and most important question in the tree. It contains the entire dataset before any splitting.

**2. Decision Nodes**

These are intermediate nodes where further questions are asked to split the data.

**3. Leaf Nodes**

The Leaf Nodes are the final outcomes or predictions. No further splitting occurs here.



**Real-Life Example: Loan Approval**

Suppose a bank wants to decide whether to approve a loan.

```text
Income > 50,000?
│
├── Yes
│     │
│     └── Credit Score > 700?
│            ├── Yes → Approve Loan
│            └── No  → Reject Loan
│
└── No → Reject Loan
```

**Identifying the Parts**

* **Root Node:** `Income > 50,000?`
* **Decision Node:** `Credit Score > 700?`
* **Leaf Nodes:**

  * Approve Loan
  * Reject Loan


1. The tree starts at the root node.
2. It checks whether the applicant's income is greater than 50,000.
3. Depending on the answer, it moves to the next branch.
4. It continues asking questions until it reaches a leaf node.
5. The leaf node gives the final prediction.


**Q2. Differentiate between Gini Impurity and Entropy. 
Which one is used by default in Scikit-learn and why?**

Both Gini Impurity and Entropy are measures of how mixed or impure the data is in a Decision Tree. The tree uses them to find the best split.


* Gini Impurity → Measures how often a randomly chosen item would be incorrectly classified.
* Entropy → Measures the amount of uncertainty or randomness in the data.

Lower values of both indicate purer groups.

**Differences**

| Gini Impurity                                   | Entropy                                       |
| ----------------------------------------------- | --------------------------------------------- |
| Simpler and faster to calculate                 | More complex and slightly slower              |
| Values range from 0 to 0.5 (for binary classes) | Values range from 0 to 1 (for binary classes) |
| Used in the CART algorithm                      | Used in ID3/C4.5 algorithms                   |
| Computationally efficient                       | Computationally more expensive                |

**Example**

Suppose a node contains:

* 50% Yes
* 50% No

This is highly mixed.

* Gini Impurity = High
* Entropy = High

If a node contains:

* 100% Yes

Then:

* Gini Impurity = 0
* Entropy = 0

This is a pure node.

**Gini Impurity** is used by default in Scikit-learn.

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()
```

The default criterion is `"gini"`.

* It is faster to compute.
* It usually produces results very similar to Entropy.
* It is computationally more efficient, especially for large datasets.



**Q3. What is Overfitting in Decision Trees?**

**How can we detect it using training and testing accuracy?**

Overfitting is a condition in which a Decision Tree learns the training data too closely, including noise and unnecessary details, instead of learning the underlying patterns. As a result, the model performs very well on the training data but fails to make accurate predictions on new, unseen data.

In Decision Trees, overfitting usually occurs when the tree becomes very deep and creates too many branches. Such a tree memorizes the training data rather than generalizing from it.

Overfitting can be detected by comparing the training accuracy and testing accuracy of the model.

For example:

| Dataset       | Accuracy |
| ------------- | -------- |
| Training Data | 99%      |
| Testing Data  | 70%      |

In this case, the training accuracy is very high, but the testing accuracy is much lower. This indicates that the model has learned the training data extremely well but is unable to perform effectively on unseen data. Therefore, the model is overfitting.

On the other hand, if the training and testing accuracies are close to each other, the model is likely to generalize well.

Example:

| Dataset       | Accuracy |
| ------------- | -------- |
| Training Data | 90%      |
| Testing Data  | 88%      |

This suggests that the model has learned the patterns in the data without memorizing them.



**Q4. Explain Pruning in Decision Trees.**

**What is the difference between Pre-pruning and Post-pruning?**

Pruning is a technique used to reduce the size and complexity of a Decision Tree by removing unnecessary branches. The main purpose of pruning is to prevent overfitting and improve the model's performance on unseen data.

When a Decision Tree grows too deep, it may start learning noise and specific details from the training data. Pruning helps simplify the tree so that it focuses only on the important patterns.

**Types of Pruning**

**1. Pre-pruning (Early Stopping)**

In Pre-pruning, the growth of the tree is stopped before it becomes too complex. Conditions are applied during tree construction to limit its growth.

Common parameters used for Pre-pruning include:

* `max_depth`
* `min_samples_split`
* `min_samples_leaf`
* `max_leaf_nodes`

Example:

```python
DecisionTreeClassifier(max_depth=3)
```

Here, the tree will not grow beyond a depth of 3.

**Advantages:**

* Reduces training time.
* Prevents the tree from becoming too large.
* Helps avoid overfitting.

**Disadvantage:**

* The tree may stop growing too early and miss important patterns.



**2. Post-pruning**

In Post-pruning, the tree is allowed to grow fully first. After the complete tree is built, unnecessary branches are removed based on their contribution to prediction performance.

The idea is to:

1. Build the full tree.
2. Remove branches that do not improve accuracy significantly.
3. Keep only the most useful splits.

**Advantages:**

* Often produces a better and more accurate tree.
* Reduces overfitting while preserving important patterns.

**Disadvantage:**

* More computationally expensive because the full tree must be built first.

**Difference Between Pre-pruning and Post-pruning**

| Pre-pruning                                             | Post-pruning                                     |
| ------------------------------------------------------- | ------------------------------------------------ |
| Stops tree growth early                                 | Builds full tree first, then removes branches    |
| Faster and less computationally expensive               | More computationally expensive                   |
| May miss useful patterns                                | Usually retains more useful patterns             |
| Applied during tree construction                        | Applied after tree construction                  |
| Uses parameters like `max_depth` and `min_samples_leaf` | Removes branches based on performance evaluation |




**Q5. What is Feature Importance?**
    
**How can it help businesses in decision-making?**

Feature Importance is a technique used to determine how much each feature (independent variable) contributes to the predictions made by a machine learning model. It helps identify which features have the greatest influence on the target variable.

In a Decision Tree, feature importance is calculated based on how much a feature reduces impurity (such as Gini Impurity or Entropy) when it is used to split the data. Features that contribute more to improving predictions are assigned higher importance scores.

**Example**

Consider a loan approval dataset with the following features:

* Income
* Credit Score
* Age
* Employment Status

After training a Decision Tree model, the feature importance scores may be:

| Feature           | Importance |
| ----------------- | ---------- |
| Credit Score      | 0.50       |
| Income            | 0.30       |
| Employment Status | 0.15       |
| Age               | 0.05       |

This indicates that Credit Score is the most important factor in determining whether a loan should be approved.

**How Feature Importance Helps Businesses**

Feature importance helps businesses understand the key factors that influence outcomes and supports better decision-making.

For example:

* **Banking:** Identifies the most important factors affecting loan approval.
* **Marketing:** Determines which customer attributes influence product purchases.
* **Healthcare:** Identifies factors that contribute most to disease prediction.
* **Human Resources:** Helps understand factors affecting employee attrition.

By focusing on the most important features, businesses can make data-driven decisions, optimize processes, reduce costs, and improve overall performance.

**Q9. Feature Selection & Splitting**

**Separate:**

**Features (X)**

**Target (y)**

**Perform train-test split (80-20)**

**Interpret:**

**Why is train-test split important?**

````python
X = df.drop('deposit',axis=1)
````

````python
y = df['deposit']
````

````python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42
)
````

Train-test split is important because it allows us to evaluate how well a machine learning model performs on unseen data.

The training set is used to train the model and learn patterns from the data. The testing set is used to evaluate the model's performance on new data that was not used during training.

Without a train-test split, the model may simply memorize the training data, leading to overfitting. In such cases, the model may show high accuracy during training but perform poorly on new data.

By keeping separate training and testing datasets, we can obtain a more reliable estimate of the model's ability to generalize to unseen observations.

**Q10. Model Building**

**Train a Decision Tree Classifier**

**Use:**

**criterion='gini'**

**max_depth=5**

**Interpret:**

**What does max_depth control?**

````python
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=5,
    random_state=42
)

dt_model.fit(X_train, y_train)
````

max_depth controls the maximum number of levels that a Decision Tree can grow.

A larger depth allows the tree to learn more complex patterns, but it can also lead to overfitting, where the model memorizes the training data instead of learning general patterns.

A smaller depth creates a simpler tree that may generalize better to new data but can sometimes lead to underfitting if important patterns are missed.

In this model, max_depth=5 restricts the tree to a maximum of 5 levels, helping to balance model complexity and performance while reducing the risk of overfitting.
